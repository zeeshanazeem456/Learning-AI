```python
contacts = [
    {"name": "Ali", "email": "Ali@gmail.com", "phone": "123"},
    {"name": "Ahmed", "email": "ahmed@gmail.com", "phone": "456"},
]

def clean_contacts(contacts):
    seen = set()
    unique_contacts = []
    lookup = {}
    for contact in contacts:
        email = contact["email"].lower().strip()
        if email in seen:
            continue
        seen.add(email)
        unique_contacts.append(contact)
        lookup[email] = contact
    return unique_contacts, lookup


def get_contact(email):
    email = email.lower().strip()
    return lookup.get(email)

unique_contacts, lookup = clean_contacts(contacts)

print(get_contact("ALI@GMAIL.COM"))
print(get_contact("unknown@GMAIL.COM"))
```
